Load English --> Marathi Model

```python
from transformers import MarianMTModel, MarianTokenizer

model_name = 'Helsinki-NLP/opus-mt-en-mr'

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)
```

Translate English Text --> Marathi Text

```python
text = [
    'This is a public information message for all citizens.'
]

inputs = tokenizer(text, return_tensors = 'pt', padding = True)
translated = model.generate(**inputs)

marathi_output = tokenizer.decode(translated[0], skip_special_tokens = True)
print(marathi_output)
```

Examples

```python
sentences = [
    "The government has announced new policies.",
    "Please maintain cleanliness in public places."
]

inputs = tokenizer(sentences, return_tensors="pt", padding=True)
outputs = model.generate(**inputs)

for en, out in zip(sentences, outputs):
    print("English:", en)
    print("Marathi:", tokenizer.decode(out, skip_special_tokens=True))
    print()
```

```python
model_name = 'sarvami/sarvam-translate'
```

```python
tokenizer = AutoProcessor.from_pretrained("sarvamai/sarvam-translate")
model = AutoModelForImageTextToText.from_pretrained("sarvamai/sarvam-translate")
```

```python
tgt_lang = 'Hindi'
input_txt = 'Be the change you wish to see in the world.'
```

```python
message = [
    {'role': 'system', 'content': f'Translate the text below to {tgt_lang}.'},
    {'role': 'user', 'content': input_txt}
]
```

```python
text = tokenizer.apply_chat_template(
    message,
    tokenize = False,
    add_generation_prompt = True
)
```

```python
print(text)
print(type(text))
```

```python
text = "Translate this sentence into Hindi."
```

```python
model_inputs = tokenizer(
    text=text,
    return_tensors="pt"
)
```

```python
print("Device:", model.device)
print("Model dtype:", model.dtype)
print("Num params (approx):", sum(p.numel() for p in model.parameters())/1e9, "B")
```

```python
# Use the keyword 'text=' so it doesn't think your string is an image
inputs = tokenizer(text=text, return_tensors="pt").to(model.device)

# The rest remains the same
outputs = model.generate(
    **inputs,
    do_sample=True,
    top_p=0.95,
    top_k=50,
    max_new_tokens=100
)
```

```python
output_ids = outputs
output_text = tokenizer.decode(output_ids, skip_special_tokens = True)

print('Input:', input_txt)
print('Translation:', output_text)
```
